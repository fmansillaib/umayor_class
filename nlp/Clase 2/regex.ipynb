**Regex** (expresión regular) es un patrón que describe un conjunto de cadenas de texto. Se puede usar para buscar, extraer o reemplazar texto en Python y otros lenguajes de programación. Es una herramienta muy poderosa para procesar texto no estructurado, como logs, tickets de soporte, transacciones bancarias, etc. Se diferencia de los métodos tradicionales de búsqueda de texto porque permite definir patrones complejos y flexibles, en lugar de buscar coincidencias exactas.

## 1. Metacaracteres esenciales

| Símbolo | Significado | Ejemplo |
|---------|-------------|--------|
| `.`     | Cualquier carácter (excepto newline) | `a.c` → "abc", "a1c" |
| `^`     | Inicio de cadena | `^Hola` |
| `$`     | Fin de cadena | `mundo$` |
| `*`     | 0 o más repeticiones | `ab*` → "a", "ab", "abb" |
| `+`     | 1 o más repeticiones | `ab+` → "ab", "abb" |
| `?`     | 0 o 1 (opcional) | `colou?r` → "color", "colour" |
| `{n,m}` | Entre n y m repeticiones | `a{2,4}` |
| `[]`    | Clase de caracteres | `[aeiou]` |
| `\|`    | OR lógico | `gato\|perro` |
| `()`    | Grupo de captura | `(ab)+` |
| `\`     | Escapa un metacarácter | `\.` = punto literal |

## 2. Secuencias especiales

| Secuencia | Significado |
|-----------|-------------|
| `\d`  | Dígito `[0-9]` |
| `\D`  | No dígito |
| `\w`  | Carácter de palabra `[a-zA-Z0-9_]` |
| `\W`  | No carácter de palabra |
| `\s`  | Espacio en blanco |
| `\S`  | No espacio en blanco |
| `\b`  | Límite de palabra |
| `\B`  | No límite de palabra |

# **1. Ejemplo**

Este ejercicio tiene le proposito de mostrar el paso a paso de cómo construir un regex para extraer información de un texto no estructurado. El objetivo es extraer la cantidad, el producto y el precio de cada línea de texto.

````python
texto = """
2x Manzanas Fuji ......... $1.990
3 kg DE TOMATE (RED) -> $2.450/kg
1   Pan_Integral [500g]  ... CLP$1.200
- 6un Huevos!! ($150 c/u)
*Leche-Entera* 1L $890.-
4,5 Kg Papas blancas ~$1.300 c/kilo
1xPaquete: Fideos #200gr $640
2  Botellas   Aceite (1L) ... $3.200 ea.
1/2 Sandía -> $1.500
10 [Yogurt] Natural 150ML $250c/u
N° items: 3 || Detergente 3Kg || $4.990
Queso*Gouda*250g==$2.350
1 - Café molido (250 g) ... $3.890
2x Jabón líquido 750ml $1.100/u
~3 Plátanos (kg) — $990
"""
````

### **Paso 1.** Cantidad

* `\d+` → uno o más dígitos
* `[.,]?` → opcionalmente un punto o coma (para decimales)
* `\d*` → más dígitos después del separador
* `|\d+/\d+` → o, alternativamente, una fracción tipo 1/2


````python
# Paso 1. Cantidad
patron_cantidad = r'(\d+[.,]?\d*|\d+/\d+)'
print(re.findall(patron_cantidad, "4,5 Kg Papas"))
````

### **Paso 2:** saltar el ruido inicial (símbolos sueltos antes del número)

* `^` → inicio de línea

* `[\s\-\*~]*` → cero o más espacios, guiones, asteriscos o virgulillas

````python
patron_ruido_inicial = r'^[\s\-\*~]*'
print(re.sub(patron_ruido_inicial, '', "  - 4,5 Kg Papas"))
````

### **Paso 3:** el producto (la parte "sucia" del medio)

* Aquí usamos un grupo no codicioso (`.+?`) porque no sabemos qué hay exactamente, solo que termina antes del precio.
* El `?` después de `.+` es clave: evita que el regex "se coma" hasta el último `$` de la línea.



````python
patron_producto = r'(.+?)'
print(re.findall(patron_producto, "4,5 Kg Papas"))
````

### **Paso 4:** el precio

* `\$` → el signo peso (escapado porque $ es metacarácter)
* `\s*` → espacios opcionales después
* `([\d.]+)` → dígitos y puntos (para capturar miles tipo 1.990)

* El precio siempre va precedido por `$` (a veces con `CLP` antes):


````python
patron_precio = r'\$\s*([\d.]+)'

print(re.findall(patron_precio, "4,5 Kg Papas $1.300 c/kilo"))
````

### **Paso 5:** unir todo

⚠️ **Nota**: con `re.VERBOSE` los espacios literales en el patrón se ignoran salvo que los escapes o uses `[ ]`. Como aquí no tengo espacios literales sueltos, está bien, pero es un punto importante para mencionar en clase.



````python
patron_completo = re.compile(
    r'^[\s\-\*~]*'              # ruido inicial
    r'(\d+[.,]?\d*|\d+/\d+)'    # grupo 1: cantidad
    r'.*?'                       # cualquier cosa (unidad, "x", etc.)
    r'(.+?)'                     # grupo 2: producto
    r'\$\s*([\d.]+)',            # grupo 3: precio
    re.VERBOSE
)
````

````python
filas = []
for linea in texto.strip().splitlines():
    m = patron_completo.search(linea)
    if m:
        cantidad, producto, precio = m.groups()
        # limpieza básica del producto
        producto = re.sub(r'[\*\[\]\(\)\.\-_#~]', ' ', producto)
        producto = re.sub(r'\s+', ' ', producto).strip()
        filas.append((cantidad, producto, precio))
    else:
        filas.append(("???", linea, "???"))  # para mostrar fallos en clase

for f in filas:
    print(f)
````

````python
# En tabla:

df = pd.DataFrame(filas, columns=["cantidad", "producto", "precio"])
print(df.to_string(index=False))
````

## **Caso 1:** Tickets de soporte

Este es el dataset: 20 tickets de soporte, cada uno con metadata (fecha, número, cliente, prioridad, estado) y un asunto en texto libre.

**Detalle importante:** varios tickets describen el mismo problema con palabras distintas. Por ejemplo, "no puedo acceder a mi cuenta", "credenciales inválidas" y "contraseña no funciona" son, en el fondo, el mismo bug. Esto es clave para entender una limitación real de regex: **no entiende sinónimos**, solo patrones de texto literal. Vamos a tener que anticipar las variantes nosotros mismos.

### **Paso 1.1** — Encontrar: filtrar tickets de prioridad ALTA

Lo más simple que se puede hacer con regex: usar `re.search` para quedarnos solo con las líneas que cumplen un patrón.


````python
for bloque in tickets.strip().split("\n\n"):
    primera_linea = bloque.split("\n")[0]
    if re.search(r'Prioridad: ALTA', primera_linea):
        print(primera_linea)
````

### **Paso 1.2** — Extraer: capturar los campos de un ticket

Ahora vamos un nivel más allá: en lugar de solo *encontrar* líneas, queremos *extraer* cada campo (fecha, número, cliente, prioridad, estado, asunto) como datos estructurados.

Para esto usamos **grupos nombrados**: `(?P<nombre>patrón)`. Esto nos permite después llamar `.groupdict()` y obtener un diccionario directamente, en lugar de tener que acordarnos del orden de los grupos.

Construcción del patrón, pieza por pieza:

```
\[(?P<fecha>\d{4}-\d{2}-\d{2}) (?P<hora>\d{2}:\d{2})\]   -> [2026-03-02 09:14]
TICKET#(?P<numero>\d+)                                       -> TICKET#10234
\| Cliente: (?P<cliente>[^|]+)                                -> | Cliente: Juan Pérez
\| Prioridad: (?P<prioridad>\w+)                              -> | Prioridad: ALTA
\| Estado: (?P<estado>\w+)                                    -> | Estado: ABIERTO
Asunto: (?P<asunto>.+)                                         -> Asunto: No puedo acceder...
```


````python
patron_ticket = re.compile(
    r'\[(?P<fecha>\d{4}-\d{2}-\d{2}) (?P<hora>\d{2}:\d{2})\] '
    r'TICKET#(?P<numero>\d+) \| '
    r'Cliente: (?P<cliente>[^|]+) \| '
    r'Prioridad: (?P<prioridad>\w+) \| '
    r'Estado: (?P<estado>\w+)\n'
    r'Asunto: (?P<asunto>.+)'
)

ejemplo = tickets.strip().split("\n\n")[0]
m = patron_ticket.search(ejemplo)
m.groupdict()
````

### **Paso 1.3** — Estructurar: tabla completa + clasificación de problemas

Aplicamos el mismo patrón a los 20 tickets y construimos un DataFrame. Luego, para responder *"¿cuál es el problema #1?"*, definimos un segundo diccionario de patrones — uno por categoría de problema — y clasificamos el campo `asunto` de cada ticket.

Esto es exactamente el punto que mencionamos antes: como las mismas categorías se describen con palabras distintas, cada categoría necesita **varias alternativas dentro del regex** (con `|`).


````python
filas = []
for bloque in tickets.strip().split("\n\n"):
    m = patron_ticket.search(bloque)
    if m:
        d = m.groupdict()
        d['cliente'] = d['cliente'].strip()
        filas.append(d)

df_tickets = pd.DataFrame(filas)
df_tickets[['numero', 'cliente', 'prioridad', 'estado']]
````

````python
categorias = {
    'acceso_cuenta':    r'acceder.*cuenta|credenciales inv|contrase\u00f1a',
    'cobro_duplicado':  r'cobr[\u00f3o].*dos veces|cargo duplicado|reembolso|monto que no reconozco',
    'app_se_cierra':    r'se cierra sola',
    'exportar_reporte': r'exportar.*reporte|bot\u00f3n de.*exportar',
}

def clasificar(asunto):
    for categoria, patron in categorias.items():
        if re.search(patron, asunto, re.IGNORECASE):
            return categoria
    return 'otro'

df_tickets['categoria'] = df_tickets['asunto'].apply(clasificar)
df_tickets[['numero', 'prioridad', 'categoria']]
````

````python
# La pregunta de negocio: ¿cuál es el problema #1 que reportan los clientes?
df_tickets['categoria'].value_counts()
````

**Lectura de negocio:** con 20 líneas de texto libre, en segundos sabemos que *"problemas de acceso a cuenta"* es la categoría más reportada. Esa es una conversación directa con el equipo de producto o soporte — no un ejercicio académico.

> **Nota pedagógica:** este enfoque de keywords con regex es rápido y explicable, pero tiene un techo: si un cliente describe el mismo problema con palabras que no anticipamos, no lo vamos a detectar. Para eso existen los enfoques de NLP basados en embeddings o modelos de clasificación entrenados — pero regex sigue siendo la herramienta correcta cuando el vocabulario es predecible y se necesita una solución rápida y transparente.

### **Paso 2.1** — Encontrar: transacciones rechazadas

````python
for linea in transacciones.strip().splitlines():
    if re.search(r'ESTADO: RECHAZADA', linea):
        print(linea)
````

### **Paso 2.2** — Extraer los campos de una transacción

````python
patron_txn = re.compile(
    r'(?P<timestamp>[\d\-T:]+) \| '
    r'TXN-(?P<id>\d+) \| '
    r'CUENTA:(?P<cuenta>[\d\-X]+) \| '
    r'(?P<tipo>DEBITO|CREDITO) \| '
    r'\$(?P<monto>[\d\.]+) \| '
    r'COMERCIO: (?P<comercio>.+?) \| '
    r'ESTADO: (?P<estado>\w+)'
)

ejemplo = transacciones.strip().splitlines()[0]
m = patron_txn.search(ejemplo)
m.groupdict()
````

### **Paso 2.3** — Estructurar: detectar cargos duplicados

Convertimos a tabla, normalizamos el monto (quitar puntos de miles y convertir a entero), y agrupamos por cuenta + comercio + monto para encontrar **cargos que se repiten exactamente** — una señal clásica de cobro duplicado o suscripción fantasma.


````python
filas = [m.groupdict() for linea in transacciones.strip().splitlines()
         if (m := patron_txn.search(linea))]

df_txn = pd.DataFrame(filas)
df_txn['monto'] = df_txn['monto'].str.replace('.', '', regex=False).astype(int)
df_txn
````

````python
# Pregunta de negocio: ¿qué cargos se repiten exactamente (mismo monto, mismo comercio, misma cuenta)?
duplicados = (
    df_txn.groupby(['cuenta', 'comercio', 'monto'])
    .size()
    .reset_index(name='veces')
)
duplicados[duplicados['veces'] > 1]
````

**Lectura de negocio:** vemos que el cargo de Netflix y la suscripción de streaming aparecen mes a mes (esperado), pero también que "Falabella Retail" se repite con el *mismo monto exacto* tres veces — vale la pena revisar si es una compra recurrente legítima o un cargo automático que no debería estar duplicándose.

### **Paso 3.1 a 3.3** — Esta vez en un solo bloque

````python
patron_log = re.compile(
    r'(?P<fecha>\d{4}-\d{2}-\d{2}) (?P<hora>\d{2}:\d{2}:\d{2}) '
    r'\[(?P<nivel>\w+)\] (?P<server>[\w\-]+) - '
    r'(?P<mensaje>.+?) - status=(?P<status>\d+)'
)

filas = [m.groupdict() for linea in logs.strip().splitlines()
         if (m := patron_log.search(linea))]

df_logs = pd.DataFrame(filas)
df_logs
````
