## 2. Pandas

[Pandas](https://pandas.pydata.org/about/index.html) es una librería que proporciona estructuras de datos y herramientas de análisis de datos de alto rendimiento y fáciles de usar. 
* La estructura de datos principal es el DataFrame, que puede considerarse como una tabla 2D en memoria (como una hoja de cálculo, con nombres de columna y etiquetas de fila). 
* Muchas funciones disponibles en Excel están disponibles mediante programación, como crear tablas dinámicas, calcular columnas basadas en otras columnas, trazar gráficos, etc.
* Proporciona un alto rendimiento para manipular (unir, dividir, modificar…) grandes conjuntos de datos

```python
import pandas as pd
```

La librería Pandas, de manera genérica, contiene las siguientes **estructuras de datos**:
* **Series**: Array de una dimensión
* **DataFrame**: Se corresponde con una tabla de 2 dimensiones
* **Panel**: Similar a un diccionario de DataFrames

```python
# Creacion de un objeto Series
s = pd.Series([2, 4, 6, 8, 10])
print(s)
```

```python
# Creación de un objeto Series inicializándolo con un diccionario de Python
altura = {"Santiago": 187, "Pedro": 178, "Julia": 170, "Ana": 165}
s = pd.Series(altura)
print(s)
```

```python
# Creación de un DataFrame inicializándolo con un diccionario de objetios Series
personas = {
    "peso": pd.Series([84, 90, 56, 64], ["Santiago","Pedro", "Ana", "Julia"]),
    "altura": pd.Series({"Santiago": 187, "Pedro": 178, "Julia": 170, "Ana": 165}),
    "hijos": pd.Series([2, 3], ["Pedro", "Julia"])
}

df = pd.DataFrame(personas)
df
```

* Acceder elementos del dataframe

```python
df["peso"]
```

```python
df[["peso", "altura"]]
```

```python
# Pueden combinarse los metodos anteriores con expresiones booleanas
df[df["peso"] > 80]
```

```python
# Pueden combinarse los metodos anteriores con expresiones booleanas
df[(df["peso"] > 80) & (df["altura"] > 180)]
```

### 2.1. Cargas de Base de Datos externos

[Read de Pandas](https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.read_csv.html) Estas funciones hacen de Pandas una herramienta extremadamente versátil para la carga y manipulación de diversos formatos de datos, permitiendo a los usuarios trabajar con una amplia variedad de fuentes de datos en análisis de datos y ciencia de datos.

1. **CSV** (`read_csv`): Lee archivos delimitados por comas (CSV) o separados por otros delimitadores, que son comunes en datos tabulares.

2. **Excel** (`read_excel`): Lee datos de archivos Excel (`.xls`, `.xlsx`), utilizando internamente paquetes como `xlrd`, `openpyxl` o `odf`.

3. **HDF5** (`read_hdf`): Lee archivos HDF5, que son formatos ideales para almacenar grandes cantidades de datos científicos.

4. **SQL** (`read_sql`): Lee datos directamente de bases de datos SQL utilizando sentencias SQL, haciendo uso de SQLAlchemy o una conexión directa a la base de datos.

5. **JSON** (`read_json`): Lee datos desde un string JSON, transformándolos directamente en un DataFrame.

6. **HTML** (`read_html`): Extrae tablas de archivos HTML, lo cual es útil para scraping de datos de páginas web.

7. **Feather** (`read_feather`): Lee archivos en formato Feather, un formato de archivo binario rápido diseñado para la interoperabilidad entre DataFrame de Python y R.

8. **Parquet** (`read_parquet`): Lee archivos en formato Parquet, que es un formato de almacenamiento en columnas optimizado para el trabajo con grandes volúmenes de datos.

9. **Msgpack** (`read_msgpack`)(Deprecated): Anteriormente se utilizaba para leer datos en el formato de serialización de Msgpack, pero ha sido deprecado desde la versión 1.0.0.

10. **Stata** (`read_stata`): Lee datos de archivos de Stata (.dta), utilizados comúnmente en investigaciones económicas y sociales.

11. **SAS** (`read_sas`): Lee datos de archivos de SAS (.sas7bdat y .xpt) que son utilizados en análisis estadístico avanzado.

12. **SPSS** (`read_spss`): Lee archivos de SPSS (.sav), un software ampliamente utilizado en la investigación social.

13. **Google BigQuery** (`read_gbq`): Extrae datos desde Google BigQuery, un servicio de almacenamiento de datos grande en la nube.

14. **Clipboard** (`read_clipboard`): Captura contenido de tu portapapeles y lo convierte en un DataFrame, útil para copiar y pegar datos directamente en Python.

15. **ORC** (`read_orc`): Lee archivos ORC (Optimized Row Columnar), que son utilizados en sistemas como Hadoop para un almacenaje eficiente de datos en columnas.

### 2.4. Funciones personalizadas sobre Dataframe

* Función personalizada para estadística 

```python
# Función para calcular la media y desviación estándar de una columna específica

def mean_and_sd(column_name):
    
    media = df[column_name].mean()
    sd = df[column_name].std()
    
    return media, sd
```

```python
# Llama a la función pasando el nombre de la columna 'price' como string
media_precio, sd_precio = mean_and_sd('price')

# Imprime los resultados
print("Media del precio:", media_precio)
print("Desviación estándar del precio:", sd_precio)
```

* Función personalizada por observación

```python
# Aplicar una función externa a una columna del DataFrame
def func(x):
    return x * 980
```

```python
df['new_price_clp'] = df["price"].apply(func)
df[['price', 'new_price_clp']]
```

```python
# Usando función lambda
# Aplicar una función lambda directamente con apply para crear una nueva columna
df['new_price_clp_2'] = df['price'].apply(lambda x: x * 980)

# Mostrar las columnas 'price' y 'new_price_clp' para ver el resultado
df[['price', 'new_price_clp']]
```

# Exportar Dataframe

Pandas también proporciona varias funciones para exportar DataFrames a diferentes formatos de archivos, lo que permite una fácil distribución y almacenamiento de datos. Te muestro una lista de las principales funciones de exportación (`to_...`) disponibles en Pandas, con una breve descripción de cada una:

1. **CSV** (`to_csv`): Exporta un DataFrame a un archivo delimitado por comas (CSV) o por otro delimitador especificado, lo cual es útil para compatibilidad con muchos sistemas.

2. **Excel** (`to_excel`): Escribe un DataFrame a un archivo Excel (`.xls`, `.xlsx`), permitiendo conservar el formato y las características propias de Excel.

3. **HDF5** (`to_hdf`): Guarda un DataFrame en un archivo HDF5, ideal para trabajar con grandes volúmenes de datos y acceso rápido a segmentos específicos de los datos.

4. **JSON** (`to_json`): Convierte un DataFrame en una cadena JSON, facilitando la integración con aplicaciones web y APIs que consumen este formato.

5. **HTML** (`to_html`): Convierte un DataFrame en una tabla HTML, útil para la visualización en páginas web y reportes basados en HTML.

6. **Markdown** (`to_markdown`): Exporta un DataFrame a formato Markdown, que es ampliamente usado en documentación y plataformas que soportan Markdown, como GitHub.

7. **Feather** (`to_feather`): Escribe un DataFrame a un archivo Feather, proporcionando una alta eficiencia en la lectura y escritura, especialmente útil para la transferencia de datos entre Python y R.

8. **Parquet** (`to_parquet`): Guarda un DataFrame en formato Parquet, optimizado para el almacenamiento en columnas y muy usado en entornos de big data como Hadoop.

9. **Msgpack** (`to_msgpack`)(Deprecated): Anteriormente usado para serializar DataFrames en el formato binario Msgpack, pero ha sido deprecado desde la versión 1.0.0.

10. **Stata** (`to_stata`): Exporta un DataFrame a un archivo de Stata (.dta), común en análisis estadístico en ciencias sociales y económicas.

11. **SAS** (`to_sas`): (No directamente disponible, pero los DataFrames pueden ser convertidos en formatos compatibles con SAS utilizando otros métodos).

12. **SPSS** (`to_spss`): Escribe un DataFrame a un archivo SPSS (.sav), útil en el ámbito de la investigación social.

13. **SQL** (`to_sql`): Guarda un DataFrame en una tabla SQL, utilizando SQLAlchemy o una conexión directa, lo cual es fundamental para integrar análisis de datos con bases de datos.

14. **Clipboard** (`to_clipboard`): Copia un DataFrame al portapapeles, lo que facilita pegar los datos en aplicaciones como Excel o editores de texto.

15. **Pickle** (`to_pickle`): Serializa un DataFrame en formato binario usando pickle, conveniente para almacenamiento temporal o transferencia de objetos Python entre procesos.

16. **Records** (`to_records`): Convierte un DataFrame a un array NumPy estructurado o a un registro, útil para ciertos tipos de manipulaciones numéricas.

17. **Latex** (`to_latex`): Exporta un DataFrame a formato LaTeX, ideal para la creación de documentos científicos y técnicos con alta calidad de tiposet.
