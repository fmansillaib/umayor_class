# 1. Variables
* Una variable es un elemento que tiene asingado un valor determinado.

## 1.1. Sintaxis de las variables

* Los nombres de las variables pueden tener cualquier longuitud y pueden consistir en letras mayúsculas y minúsculas (A-Z, a-z), dígitos (0-9) y el carácter de subrayado (_)

# 2. Strings

* Un string se corresponde con un conjunto de caracteres que forman una cadena de texto 

# 3. Indexación

* Python puede acceder a los caracteres de un conjunto ordenado de datos mediante un índice. 
* el string `Hola` -> `[0] = H` ; `[1] = o` ; `[2] = l` ; `[3] = a`

# 4. Slicing

* Al aplicar una sintaxis especifica de indexación se podrá extraer subcadenas de una cadena texto.
* Una cadena `s` es de la forma `s[m:n]` 

# 5. Stride

* Si se añade un `:`adicional y un tercer indice, se designa un stride, que indica cuantos caracteres saltar hasta obtener el siguiente caracter

palabra = 'Pan & Miel'

# 6. Números

* Existen tres tipos de datos numéricos: enteros (`int`), números de punto flotante (`float`) y números complejos

* Para números grande no se separa por `.` o por `,` se separan por `_`

* Los números de punto flotante son aquellos que tiene una parte decimal, se relacionan con `float`
