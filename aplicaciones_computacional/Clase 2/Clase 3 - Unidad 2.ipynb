# 3. Indexación

* Python puede acceder a los caracteres de un conjunto ordenado de datos mediante un índice. 
* el string `Hola` -> `[0] = H` ; `[1] = o` ; `[2] = l` ; `[3] = a`

```python
palabra = 'Taza'
```

```python
# Reordenar la palabra a Zata
print(palabra[2], palabra[1], palabra[0], palabra[3])
```

```python
# Se puede indezar con números negativos para indexar por detras del objeto. 
 
print(palabra[-1], palabra[-2], palabra[-3], palabra[-4])
```

```python
# Error 1.

palabra[0] = 's'
```

# 4. Slicing

* Al aplicar una sintaxis especifica de indexación se podrá extraer subcadenas de una cadena texto.
* Una cadena `s` es de la forma `s[m:n]` 

```python
palabra = 'Pan & Miel'
```

```python
palabra[0:3]
```

```python
palabra[-4:]
```

```python
palabra[6:]
```

# 5. Stride

* Si se añade un `:`adicional y un tercer indice, se designa un stride, que indica cuantos caracteres saltar hasta obtener el siguiente caracter

palabra = 'Pan & Miel'

```python
palabra[::]
```

```python
palabra[0:3]
```

```python
palabra[0:3:2]
```

```python
palabra[::2]
```
