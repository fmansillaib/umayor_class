```python
# ==============================================================
# Actividad de Estadística – Búsqueda de hoja
# Integrantes: [Nombre 1] y [Nombre 2]
# ==============================================================

def buscar_hoja(medidas):
    """
    Recibe una lista de longitudes (en cm) de hojas.
    El programa termina cuando se encuentra una hoja
    que mide entre 6.8 y 7.2 cm (inclusive).
    """

    # Definimos los rangos
    rango_interes = [5.0, 8.0]     # Hojas normales
    rango_objetivo = [6.8, 7.2]    # Hoja buscada

    # Creamos contadores
    total = 0
    dentro = 0
    fuera = 0
    suma = 0
    minimo = None
    maximo = None

    # Recorremos cada medida
    for hoja in medidas:
        total += 1  # contamos la hoja

        # ¿Está dentro del rango de interés?
        if rango_interes[0] <= hoja <= rango_interes[1]:
            dentro += 1
            suma += hoja
            # guardamos la más chica y la más grande
            if minimo is None or hoja < minimo:
                minimo = hoja
            if maximo is None or hoja > maximo:
                maximo = hoja
        else:
            fuera += 1

        # Si está en el rango objetivo, terminamos
        if rango_objetivo[0] <= hoja <= rango_objetivo[1]:
            break

    # Calculamos el promedio
    if dentro > 0:
        promedio = suma / dentro
    else:
        promedio = 0

    # Mostramos los resultados
    print("====== RESULTADOS ======")
    print("1) Total de hojas medidas:", total)
    print("2) Dentro del rango (5 a 8 cm):", dentro)
    print("3) Fuera del rango:", fuera)
    print("4) Promedio (dentro del rango):", round(promedio, 2), "cm")
    print("5) Hoja más larga:", maximo, "cm")
    print("6) Hoja más corta:", minimo, "cm")
    print("========================")


# Ejemplo de uso:
medidas = [4.5, 5.2, 8.1, 6.4, 7.0, 7.3]  # se detiene en 7.0
buscar_hoja(medidas)
```
